# Distribution of the Columns/Features

Viewing the distribution of each feature helps us understand the range, common values, and whether the data is skewed or has outliers. This is important for deciding how to analyze or clean the columns.

The cell below creates a series of histograms. A histogram shows how data values are spread out by grouping them into bins and counting how many values fall into each bin.

```python
# Select only the numeric columns from the dataset
numeric_cols = df.select_dtypes(include=['number']).columns

# Define how many plots to show per row in the grid
num_cols = 3

# Calculate the number of rows needed based on total numeric columns
num_rows = (len(numeric_cols) + num_cols - 1) // num_cols

# Create a figure and a grid of subplots with the calculated size
fig, axes = plt.subplots(num_rows, num_cols, figsize=(15, num_rows * 4))

# Flatten the 2D array of axes to simplify iteration
axes = axes.flatten()

# Loop over each numeric column and corresponding subplot axis
for i, col in enumerate(numeric_cols):
    ax = axes[i]
    # Plot a histogram for the current column with 30 bins
    df[col].hist(ax=ax, bins=6, edgecolor='black')
    # Set title and axis labels for clarity
    ax.set_title(f'Distribution of {col}')
    ax.set_xlabel(col)
    ax.set_ylabel('Frequency')

# If there are unused subplot axes (in case total plots < grid size), remove them
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

# Adjust layout to prevent overlap and improve spacing
plt.tight_layout()

# Show the complete plot grid
plt.show()
```

## Distribution viewed through Boxplots

Boxplots are another way that we can view the distribution of our data, making it easy to spot the median and spread (interquartile range). This grid has boxplots for each of the features.

```python
# Select numeric columns
numeric_cols = df.select_dtypes(include=['number']).columns

# Set number of plots per row
num_cols = 3
num_rows = (len(numeric_cols) + num_cols - 1) // num_cols

# Create subplots grid
fig, axes = plt.subplots(num_rows, num_cols, figsize=(15, num_rows * 4))
axes = axes.flatten()

# Plot filled boxplots for each numeric column
for i, col in enumerate(numeric_cols):
    ax = axes[i]
    sns.boxplot(y=df[col], ax=ax, color='darkgoldenrod', width=.5)  # Use a fill color here
    ax.set_title(f'Box Plot of {col}')
    ax.set_xlabel('')
    ax.set_ylabel(col)

# Remove any unused subplots
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

## Boxplot by Grouped Target Categories

Now, let's add this information to our box plot grid.

```python
# Define your target column name
target_col = 'target'

# Make sure target is string if your palette uses string keys
df[target_col] = df[target_col].astype(str)

# Select numeric columns excluding the target
numeric_cols = df.select_dtypes(include=['number']).columns

# Set up grid layout
num_cols = 3
num_rows = (len(numeric_cols) + num_cols - 1) // num_cols
fig, axes = plt.subplots(num_rows, num_cols, figsize=(15, num_rows * 4))
axes = axes.flatten()

# Plot side-by-side boxplots
for i, col in enumerate(numeric_cols):
    ax = axes[i]
    sns.boxplot(
        x=target_col,
        y=col,
        data=df,
        hue=target_col,  # <-- assign x to hue
        palette={'0': 'red', '1': 'yellow'},
        legend=False,
        ax=ax,
        width=.5
    )
    ax.set_title(f'Box Plot of {col} by Target')
    ax.set_xlabel('Target')
    ax.set_ylabel(col)

# Remove empty subplots
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.suptitle('Box Plots of Numeric Features by Target')
plt.tight_layout()
plt.show()
```

# Independent T-Test

Prior to this, we've been *eyeballing* differences between groups, lets change that. An **independent t-test** is a simple statistical test that checks if the average value of a column/feature is really different between two target groups, or if that difference could just be random chance.

As a general rule of thumb, if you receive a p-value less than `0.05`, the difference is considered "statistically significant."

```python
# Define your tested feature
feature_to_test = "priority_level"
#task: -33.185
#cpu: - 81.709
#memory: .806
#network: .104
#io: -.74
#disk: -28.789
#connections: -1.085
#priority: -31.971
# Cast target back to an int
target = df['target'].astype(int)

# Group your feature by target
group0 = df.loc[target == 0, feature_to_test]
group1 = df.loc[target == 1, feature_to_test]

# Perform an independent t-test
t_stat, p_value = stats.ttest_ind(group0, group1)

# Print your results
print(f"\nt-statistic:    {t_stat:.3f}")
print(f"p-value:        {p_value:.10f}")
if p_value < 0.05:
    print("\nStatistically significant difference (p < 0.05).")
else:
    print("\nNo statistically significant difference (p ≥ 0.05).")
```
